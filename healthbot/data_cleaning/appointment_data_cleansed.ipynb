```python
from pyspark.sql.functions import *
from dateutil import parser

# ---------------- CONFIG ----------------
CATALOG = "chatbot_dev"
BRONZE_SCHEMA = "bronze"
SILVER_SCHEMA = "silver"

# Ensure silver schema exists
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")

df = spark.read.table(f"{CATALOG}.{BRONZE_SCHEMA}.appointments_data")

clean_date_format = (
    df.withColumn("appointment_date", try_to_date(col("appointment_date")))
      .drop("_rescued_data")
)

# Removes duplicates
clean_dedup = clean_date_format.dropDuplicates(["patient_id"])

(clean_dedup.write
  .format("delta")
  .mode("overwrite")
  .option("overwriteSchema", "true")
  .saveAsTable(f"{CATALOG}.silver.appointments_data")
)
```
